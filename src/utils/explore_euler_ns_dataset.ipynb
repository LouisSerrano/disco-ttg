```python
from euler_ns_dataset import EulerDiffusionDataset
import matplotlib.pyplot as plt
```

```python
file_dir = "/mnt/home/lserrano/ceph/data/euler_ns_short/"
num_gpus = 1
train_ds = EulerDiffusionDataset(file_dir, num_gpus, split= 'train')
```

```python
idx = 1191

traj, label = train_ds.__getitem__(idx)
```

```python
fig, axs = plt.subplots(1, 10, figsize=(20, 4))
sub_x = 1
for t in range(10):
    axs[t].imshow(traj[40+t, ::sub_x, ::sub_x])
    # Remove the ticks and box
    axs[t].axis('off')
    
    # Add a small title for context
    axs[t].set_title(f"t = {t}", fontsize=12)
fig.suptitle(f"label: {label}")
```
